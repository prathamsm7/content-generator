```python
from youtube_transcript_api import YouTubeTranscriptApi, NoTranscriptFound
```

```python

def get_transcript(video_id: str):
    """Robust transcript fetch: prefer en/en-US, else fallback to any."""
    ytt_api = YouTubeTranscriptApi()
    transcript_list = ytt_api.list(video_id)

    try:
        return transcript_list.find_transcript(["en", "en-US"]).fetch()
    except NoTranscriptFound:
        available = list(transcript_list)
        if not available:
            raise
        return available[0].fetch()


video_id = "NNH-RLNyzoM"  # change this
transcript = get_transcript(video_id)
```

```python
import yt_dlp

url = "https://www.youtube.com/watch?v={}".format(video_id)

print(url)
ydl_opts = {
    'quiet': True,
    'skip_download': True, # We only want info, not the file
    'force_generic_extractor': False
}

info = {}
with yt_dlp.YoutubeDL(ydl_opts) as ydl:
    info = ydl.extract_info(url, download=False)
    
    print(f"Title: {info.get('title')}")
    print(f"Views: {info.get('view_count')}")
    print(f"Upload Date: {info.get('upload_date')}") # Format: YYYYMMDD
    print(f"Description: {info.get('description')}") # Format: YYYYMMDD
    print(f"category: {info.get('categories')}") # Format: YYYYMMDD
    print(f"tags: {info.get('tags')}") # Format: YYYYMMDD
    print(f"chapters: {info.get('chapter')}") # Format: YYYYMMDD
```

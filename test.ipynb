```python
from youtube_transcript_api import YouTubeTranscriptApi, NoTranscriptFound
```

```python

def get_transcript(video_id: str):
    """Robust transcript fetch: prefer en/en-US, else fallback to any."""
    ytt_api = YouTubeTranscriptApi()
    transcript_list = ytt_api.list(video_id)

    try:
        return transcript_list.find_transcript(["en", "en-US"]).fetch()
    except NoTranscriptFound:
        available = list(transcript_list)
        if not available:
            raise
        return available[0].fetch()


video_id = "MPCVGFvgVEQ"  # change this
transcript = get_transcript(video_id)
```

```python
transcript
```

```python
full_transcript = []
for snippet in transcript.to_raw_data():
    full_transcript.append({"text":snippet['text']})

# print(full_transcript)
full_text = " ".join([t["text"] for t in transcript.to_raw_data()])
print(full_text)
```
